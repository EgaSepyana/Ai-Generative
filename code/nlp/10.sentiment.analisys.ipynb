```python
from textblob import TextBlob
```

```python
sentence_1 = "i had a great time at the movie it was really funny"
sentence_2 = "i had a great time at the movie but the parking was terrible"
sentence_3 = "i had a great time at the movie but the parking wasn't great"
sentence_4 = "i went to see a movie"
```

```python
sentence_1
```

```python
senttiment_score_1 = TextBlob(sentence_1)
print(sentence_1)
senttiment_score_1.sentiment.polarity
```

```python
senttiment_score_2 = TextBlob(sentence_2)
print(sentence_2)
senttiment_score_2.sentiment.polarity
```

```python
senttiment_score_3 = TextBlob(sentence_3)
print(sentence_3)
senttiment_score_3.sentiment.polarity
```

```python
senttiment_score_4 = TextBlob(sentence_4)
print(sentence_4)
senttiment_score_4.sentiment.polarity
```
